```python
import requests
from datetime import timedelta, datetime, date, tzinfo, UTC, timezone
from zoneinfo import ZoneInfo

"""
Get Current Date/Time
API Call from Date - 6 (00:00:00) to Date - 1 (23:59:59)
"""
```

```python
curr_datetime = datetime.now(ZoneInfo('America/New_York'))
end_date = (curr_datetime - timedelta(days=1)).replace(hour=23, minute=59, second=59)
start_date = (curr_datetime - timedelta(days=6)).replace(hour=0, minute=0, second=0)
```

```python
start_date.strftime('%FT%T%:z').replace(':', '%3A').replace('+', '%2B')
```

```python
end_date.strftime('%FT%T%:z').replace(':', '%3A').replace('+', '%2B')
```

```python
# Replace Colons (:) with %3A and Pluses (+) with %2B
URL = f'https://api.weather.gov/stations/KTPA/observations?start={start_date.strftime('%FT%T%:z').replace(':', '%3A').replace('+', '%2B')}&end={end_date.strftime('%FT%T%:z').replace(':', '%3A').replace('+', '%2B')}&limit=200'
noaa_response = requests.get(URL)
```
